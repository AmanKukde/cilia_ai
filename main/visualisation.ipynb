```python
import tifffile as tf
from pathlib import Path
from utils.common import *
from utils.plotting import *
```

```python
preds_path = Path("/group/jug/aman/Cilia_Ai_Predictions/Normalised_mean_std/")
data_path = Path("../data/DP/max_projection.tif")

predictions = tf.imread(preds_path/"predictions_cilia.tif")
imgs = tf.imread(data_path)
```

```python
predictions, _ = binarize_and_correct_tiff(predictions)
```

```python
print("01. Thresholding Cilia Channel based on the Green Protein Channel (1) and Red (2) Channel")
predictions_cleaned = keep_connected_blobs_union(predictions, target_channel= 3, reference_channels=(1,2))
print("\n02. Thresholding the Red (2) Channel using the Green Protein Channel (1) and Cleaned Cilia Channel (3)")
predictions_cleaned = keep_connected_blobs_union(predictions_cleaned, target_channel= 2, reference_channels=(1,3))
print("\n03. Computing the intersection of Green Protein Channel (1) and other two channels ")
predictions_cleaned[...,1] = intersect_channels(predictions_cleaned)
print("\nDone ✅")
```

```python
import ipywidgets as widgets
from IPython.display import display

# Update function
def update(idx):
    plot_input_and_predictions_with_outline(
        idx=idx,
        image=imgs,
        predictions=predictions,
        predictions_cleaned=predictions_cleaned,
        channels=[0,1,2,3],
        highlight_rows=[2],             # highlight Predictions & Mask Outlines
        highlight_colors=['red']   # blue for row 1, red for row 3
    )


# Create a bigger slider
slider = widgets.IntSlider(
    value=0,
    min=0,
    max=len(imgs)-1,
    step=1,
    description='Index:',
    continuous_update=False,
    style={'description_width': '160px'},   # make description wider
    layout=widgets.Layout(width='60%')     # make slider longer
)

# Link slider to update function
widgets.interact(update, idx=slider)
```
